#### Armado de las tablas en ARS

```python
def ars():
    Honorarios = honorarios() #DataFrame de Honorarios.
    Honorarios_ARS=Honorarios.loc[Honorarios['Moneda'] == 'Pesos'] #Filtra para quedarme con los pesos.
    Honorarios_ARS=Honorarios_ARS.sort_values('Total', ascending=False) #Ordena por Total.
    Honorarios_ARS.loc['Total'] = Honorarios_ARS[['Gastos','Honorarios','Total']].sum() #Agrega los totales a las columnas.
    Honorarios_ARS= Honorarios_ARS.replace([np.nan], '') #Elimina los NaN.
    Honorarios_ARS=Honorarios_ARS.round(2)#Redondea en dos decimales.

    #Tabla para exportar
    Honorarios_ARS_file=Honorarios_ARS[['Moneda','ID PS','Inicio período','Fin período','Vto. Factura','CNV','Fondo','Honorarios','Gastos','Total']]#Selecciono las columnas a usar
    Honorarios_ARS_file=Honorarios_ARS_file.astype(str) #Pasa todo a string.

    #Armo la tabla de pesos para el mail
    Honorarios_ARS_mail=Honorarios_ARS[['Número','Depositante','Comitente','Fondo','Cta. Cte. FCI','Cta. Cte. SISA','Moneda','Honorarios','Gastos','Total']].copy()#Selecciono las columnas a usar
    #Formato de las columnas numércias
    Honorarios_ARS_mail['Honorarios'] = Honorarios_ARS_mail['Honorarios'].apply(lambda x: '{:,.2f}'.format(x).replace(',', 'X').replace('.', ',').replace('X', '.'))
    Honorarios_ARS_mail['Gastos'] = Honorarios_ARS_mail['Gastos'].apply(lambda x: '{:,.2f}'.format(x).replace(',', 'X').replace('.', ',').replace('X', '.'))
    Honorarios_ARS_mail['Total'] = Honorarios_ARS_mail['Total'].apply(lambda x: '{:,.2f}'.format(x).replace(',', 'X').replace('.', ',').replace('X', '.'))
    
    return {"Honorarios_ARS_file": Honorarios_ARS_file, "Honorarios_ARS_mail": Honorarios_ARS_mail}
```

#### Armado de las tablas en USD

```python
def usd():
    TC = tc() #DataFrame de tipo de cambio.
    
    Honorarios = honorarios() #DataFrame de Honorarios.
    Honorarios_USD=Honorarios.loc[Honorarios['Moneda'] == 'Dólares'] #Filtra para quedarme con los dólares.
    Honorarios_USD=Honorarios_USD.rename (columns={'Honorarios':'Honorarios (USD)','Gastos':'Gastos (USD)','Total':'Total (USD)' }) #Renombra las columnas.
    pos_columna_a_insertar2=Honorarios_USD.columns.get_loc('Total (USD)') #Busca la ubicación de la columna para tener de referencia.
    Honorarios_USD.insert(pos_columna_a_insertar2+1,'Tipo de cambio',TC) #Inserta la columna con el tipo de cambio.
    Honorarios_USD.insert(pos_columna_a_insertar2+2,'Honorarios (ARS)',(Honorarios_USD['Honorarios (USD)']*TC).round(2)) #Inserta la columna y calcula los honorarios en pesos.
    Honorarios_USD.insert(pos_columna_a_insertar2+3,'Gastos (ARS)',(Honorarios_USD['Gastos (USD)']*TC).round(2)) #Inserta la columna y calcula los gastos en pesos.
    Honorarios_USD.insert(pos_columna_a_insertar2+4,'Total (ARS)',(Honorarios_USD['Total (USD)']*TC).round(2)) #Inserta la columna y calcula el total en pesos.
    Honorarios_USD=Honorarios_USD.sort_values('Total (ARS)', ascending=False) #Ordena por Total.
    Honorarios_USD.loc['Total'] = Honorarios_USD[['Gastos (USD)','Honorarios (USD)','Total (USD)']].sum() #Agrega los totales a las columnas.
    Honorarios_USD= Honorarios_USD.replace([np.nan], '') #Elimina los NaN.
    Honorarios_USD=Honorarios_USD.round(2)#Redondea en dos decimales.
    
    #Tabla para exportar
    Honorarios_USD_file=Honorarios_USD[['Moneda','ID PS','Inicio período','Fin período','Vto. Factura','CNV','Fondo','Honorarios (USD)','Gastos (USD)','Total (USD)','Tipo de cambio','Honorarios (ARS)','Gastos (ARS)','Total (ARS)']]#Selecciono las columnas a usar
    Honorarios_USD_file=Honorarios_USD_file.round(2)#Redondea en dos decimales
    
    
    Honorarios_USD_file=Honorarios_USD_file.astype(str) #Pasa todo a string

    #Armo la tabla de dólares para el mail
    Honorarios_USD_mail=Honorarios_USD[['Número','Depositante','Comitente','Fondo','Cta. Cte. FCI','Cta. Cte. SISA','Moneda','Honorarios (USD)','Gastos (USD)','Total (USD)']].copy()#Selecciono las columnas a usar
    #Formato de las columnas numércias
    Honorarios_USD_mail['Honorarios (USD)'] = Honorarios_USD_mail['Honorarios (USD)'].apply(lambda x: '{:,.2f}'.format(x).replace(',', 'X').replace('.', ',').replace('X', '.'))
    Honorarios_USD_mail['Gastos (USD)'] = Honorarios_USD_mail['Gastos (USD)'].apply(lambda x: '{:,.2f}'.format(x).replace(',', 'X').replace('.', ',').replace('X', '.'))
    Honorarios_USD_mail['Total (USD)'] = Honorarios_USD_mail['Total (USD)'].apply(lambda x: '{:,.2f}'.format(x).replace(',', 'X').replace('.', ',').replace('X', '.'))

    return{"Honorarios_USD_file":Honorarios_USD_file,"Honorarios_USD_mail":Honorarios_USD_mail}
```

#### Exportación de los archivos

```python
def export():
    tablasARS = ars() #Convierte en un Dataframe la salida de la función.
    tablasUSD = usd()  #Convierte en un Dataframe la salida de la función.
    
    ARS = tablasARS["Honorarios_ARS_file"] #Convierte la tabla en un Dataframe.
    USD = tablasUSD["Honorarios_USD_file"] #Convierte la tabla en un Dataframe.

    writer = pd.ExcelWriter("V:/BIVSA/ADMOPER/Facturacion/Honorarios FCI-" + fa + ".xlsx")
    ARS.to_excel(writer, sheet_name="Facturas ARS", index=False)
    USD.to_excel(writer, sheet_name="Facturas USD", index=False)
    writer.close()

    print('El archivo se generó correctamente en V:\BIVSA\ADMOPER\Facturacion.')
```
