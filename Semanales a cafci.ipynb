```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import datetime as dt
import xml.etree.ElementTree as ET
import os
```

```python
# Filtrar XML

def Filtrar_XML (archivo_entrada,archivo_salida): 
    try:
        tree=ET.parse(archivo_entrada)
        root=tree.getroot()
        umbral=0.001
        activos_a_eliminar=[]
        for Activo in root.findall(".//Activo"): 
            Cant=Activo.find("Cant")
            Cant=float(Cant.text)
            if abs(Cant) < umbral:
                activos_a_eliminar += [Activo]
                print(Cant)
        for Activo in activos_a_eliminar:
            for parent in root.findall(".//Activo/.."):
                if Activo in parent:
                    parent.remove(Activo)
                    break
        tree.write(archivo_salida, encoding="utf-8",xml_declaration=True)
        print(f"Proceso finalizado")
    except Exception as E:
        print(f"Error al procesar XML{E}")

```

```python
for f in Files: 
    if f.endswith(".xml"):
        print (path+f)
        Filtrar_XML(path+f,path+f)
```
